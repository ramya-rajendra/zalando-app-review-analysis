```python
from google_play_scraper import reviews, Sort
import pandas as pd

result, _ = reviews(
    'de.zalando.mobile',
    lang='de', country='de',
    sort=Sort.NEWEST,
    count=3000
)

df = pd.DataFrame(result)
df = df[['reviewId', 'content', 'score', 'thumbsUpCount', 'appVersion', 'at']]
df = df.dropna(subset=['content'])

df.to_csv('../data/zalando_reviews.csv', index=False)
print(len(df))
df.head()
```

```python
for i in range(0, len(df), 200):
    df.iloc[i:i+200][['reviewId', 'content']].to_csv(
        f'../data/part_{i//200+1}.csv', index=False)

print("Created", (len(df) + 199) // 200, "files")
```
